```python
import numpy as np
import pandas as pd

def generate_and_save_data(distribution_type, n_samples, filename):

    if distribution_type == 'uniform':
        data = np.random.uniform(low=0, high=1, size=n_samples)
    elif distribution_type == 'gaussian':
        data = np.random.normal(loc=0, scale=1, size=n_samples)
    elif distribution_type == 'skewed':
        # Using log-normal distribution for skewed data
        data = np.random.lognormal(mean=0, sigma=1, size=n_samples)
    else:
        raise ValueError("Invalid distribution_type. Choose from 'uniform', 'gaussian', or 'skewed'.")

    df = pd.DataFrame({'value': data})
    df.to_csv(filename, index=False)
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    return merge(left_half, right_half)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def quick_sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)

def partition(arr, low, high):
    i = (low - 1)
    pivot = arr[high]
    for j in range(low, high):
        if arr[j] <= pivot:
            i = i + 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return (i + 1)
```

## Módulo 2 - algoritmos

### Subtask:
Modificar los algoritmos de ordenamiento para incluir contadores de comparaciones.


```python
def bubble_sort(arr):
    n = len(arr)
    comparison_count = 0
    for i in range(n):
        for j in range(0, n - i - 1):
            comparison_count += 1 # Increment counter for comparison
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr, comparison_count

def selection_sort(arr):
    n = len(arr)
    comparison_count = 0
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparison_count += 1 # Increment counter for comparison
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr, comparison_count

def insertion_sort(arr):
    comparison_count = 0
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            comparison_count += 2 # Increment counter for both parts of the while condition
            arr[j + 1] = arr[j]
            j -= 1
        if j >= 0: # Check the first part of the condition one last time if the while loop was never entered
             comparison_count += 1
        arr[j + 1] = key
    return arr, comparison_count

def merge_sort(arr):
    if len(arr) <= 1:
        return arr, 0

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    left_half, left_comparisons = merge_sort(left_half)
    right_half, right_comparisons = merge_sort(right_half)

    merged_arr, merge_comparisons = merge(left_half, right_half)
    total_comparisons = left_comparisons + right_comparisons + merge_comparisons

    return merged_arr, total_comparisons

def merge(left, right):
    result = []
    i = j = 0
    comparison_count = 0
    while i < len(left) and j < len(right):
        comparison_count += 1 # Increment counter for comparison between left and right elements
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result, comparison_count

def quick_sort(arr, low, high):
    comparison_count = 0
    if low < high:
        pi, partition_comparisons = partition(arr, low, high)
        comparison_count += partition_comparisons
        _, left_comparisons = quick_sort(arr, low, pi - 1)
        _, right_comparisons = quick_sort(arr, pi + 1, high)
        comparison_count += left_comparisons + right_comparisons

    return arr, comparison_count

def partition(arr, low, high):
    i = (low - 1)
    pivot = arr[high]
    comparison_count = 0
    for j in range(low, high):
        comparison_count += 1 # Increment counter for comparison with pivot
        if arr[j] <= pivot:
            i = i + 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return (i + 1), comparison_count
```

### Subtarea:
Implementar algoritmos de búsqueda lineal y búsqueda binaria.

```python
def busqueda_lineal(lista, valor_objetivo):
    """
    Implementa la búsqueda lineal en una lista.

    Args:
        lista: La lista en la que buscar.
        valor_objetivo: El valor a buscar.

    Returns:
        El índice del primer elemento que coincide con el valor objetivo, o -1 si no se encuentra.
    """
    for indice, elemento in enumerate(lista):
        if elemento == valor_objetivo:
            return indice
    return -1

def busqueda_binaria(lista_ordenada, valor_objetivo):
    """
    Implementa la búsqueda binaria en una lista ordenada.

    Args:
        lista_ordenada: La lista ordenada en la que buscar.
        valor_objetivo: El valor a buscar.

    Returns:
        El índice del elemento que coincide con el valor objetivo, o -1 si no se encuentra.
    """
    izquierda, derecha = 0, len(lista_ordenada) - 1
    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2
        if lista_ordenada[medio] == valor_objetivo:
            return medio
        elif lista_ordenada[medio] < valor_objetivo:
            izquierda = medio + 1
        else:
            derecha = medio - 1
    return -1
```

Cree un decorador para medir el tiempo de ejecución de las funciones.

```python
import time

def medir_tiempo_ejecucion(func):
    """
    Decorador para medir el tiempo de ejecución de una función.
    """
    def wrapper(*args, **kwargs):
        tiempo_inicio = time.time()
        resultado = func(*args, **kwargs)
        tiempo_fin = time.time()
        duracion = tiempo_fin - tiempo_inicio
        print(f"La función '{func.__name__}' tardó {duracion:.6f} segundos en ejecutarse.")
        return resultado
    return wrapper
```

```python
# Assuming Producto class and quick_sort function are already defined in the environment

def quick_sort_productos(arr, low, high, key):
    """
    Implementa la clasificación rápida en una lista de objetos Producto
    basándose en un atributo clave.

    Args:
        arr: La lista de objetos Producto a ordenar.
        low: El índice inicial del subarreglo.
        high: El índice final del subarreglo.
        key: El atributo del objeto Producto a usar para la comparación (por ejemplo, 'precio' o 'nombre').
    """
    if low < high:
        pi = partition_productos(arr, low, high, key)
        quick_sort_productos(arr, low, pi - 1, key)
        quick_sort_productos(arr, pi + 1, high, key)

def partition_productos(arr, low, high, key):
    """
    Función auxiliar para quick_sort que realiza la partición de la lista
    basándose en un atributo clave de los objetos Producto.

    Args:
        arr: La lista de objetos Producto a particionar.
        low: El índice inicial del subarreglo.
        high: El índice final del subarreglo.
        key: El atributo del objeto Producto a usar para la comparación.

    Returns:
        El índice de partición.
    """
    pivot = getattr(arr[high], key)
    i = (low - 1)
    for j in range(low, high):
        if getattr(arr[j], key) <= pivot:
            i = i + 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return (i + 1)

def ordenar_por_precio(lista_productos):
    """
    Ordena una lista de objetos Producto por precio utilizando quick sort.

    Args:
        lista_productos: La lista de objetos Producto a ordenar.

    Returns:
        Una nueva lista de objetos Producto ordenada por precio.
    """
    # Crear una copia para no modificar la lista original
    lista_copia = lista_productos[:]
    quick_sort_productos(lista_copia, 0, len(lista_copia) - 1, 'precio')
    return lista_copia

def ordenar_por_nombre(lista_productos):
    """
    Ordena una lista de objetos Producto por nombre utilizando quick sort.

    Args:
        lista_productos: La lista de objetos Producto a ordenar.

    Returns:
        Una nueva lista de objetos Producto ordenada por nombre.
    """
    # Crear una copia para no modificar la lista original
    lista_copia = lista_productos[:]
    quick_sort_productos(lista_copia, 0, len(lista_copia) - 1, 'nombre')
    return lista_copia

# Assuming 'inventario' list is already generated

# Ordenar por precio y mostrar los primeros 5 productos
inventario_ordenado_precio = ordenar_por_precio(inventario)
print("Inventario ordenado por precio (primeros 5):")
for i in range(min(5, len(inventario_ordenado_precio))):
    print(inventario_ordenado_precio[i])

# Ordenar por nombre y mostrar los primeros 5 productos
inventario_ordenado_nombre = ordenar_por_nombre(inventario)
print("\nInventario ordenado por nombre (primeros 5):")
for i in range(min(5, len(inventario_ordenado_nombre))):
    print(inventario_ordenado_nombre[i])
```
